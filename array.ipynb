```python
## using recursion

def countDigits(num):
  if num == 0:  ## check if num = 0, then return 0
    return 0
  num = num//10 ## remove the last digit of num  
  return 1 + countDigits(num)  ## 1(count for last digit) + recursive call 

countDigits(1232)
```

```python
## using iteration

def countDigitsIter(num):
  count = 0  ## initialize count with 0 
  while num > 0: ## check if num is greater than 0
    num = num//10  ## remove the last digit from num
    count += 1 ## increment the count by 1 after every iteration that is for every removed digit
  return count

countDigitsIter(12324)
```

```python
def sumOfDigits(num):
  sum = 0
  while num > 0:
    digit = num%10
    sum += digit
    num = num//10
  return sum

sumOfDigits(113)
```
